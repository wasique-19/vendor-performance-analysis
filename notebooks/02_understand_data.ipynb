```python
import pandas as pd

vendors = pd.read_csv("../data/raw/vendors.csv")
products = pd.read_csv("../data/raw/products.csv")
orders = pd.read_csv("../data/raw/purchase_orders.csv")
deliveries = pd.read_csv("../data/raw/deliveries.csv")
```

```python
for name, df in [("vendors", vendors), ("products", products),
                 ("orders", orders), ("deliveries", deliveries)]:
    print(f"\n=== {name} ===")
    df.info()
```

```python
orders[["quantity_ordered", "unit_price", "total_amount"]].describe().round(1)
```

```python
print(orders.nunique())                       # distinct values per column
print("Duplicate po_id:", orders["po_id"].duplicated().sum())
print(orders["po_status"].value_counts())
```

```python
print(orders["vendor_id"].isin(vendors["vendor_id"]).all())
print(orders["product_id"].isin(products["product_id"]).all())
print(deliveries["po_id"].isin(orders["po_id"]).all())
print("Unique POs in deliveries:", deliveries["po_id"].nunique())
```
